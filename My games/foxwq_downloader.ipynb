```python
%load_ext autoreload
%autoreload 2

import os, time
from fox_full_downloader import fetch_chess_list, fetch_all_games, fetch_sgf, make_filename

# 先看看第一页有几条
payload = fetch_chess_list("531169471", lastcode=0)
print(f"第一页 {len(payload['chesslist'])} 条")
print(f"第一条: {payload['chesslist'][0]['blackenname']} VS {payload['chesslist'][0]['whiteenname']}")

# 拉全部
games = fetch_all_games("531169471")
print(f"\n总共 {len(games)} 局")

# 看一下时间跨度
print(f"最近一局: {games[0]['starttime']}")
print(f"最早一局: {games[-1]['starttime']}")

output_dir = os.path.expanduser("~/Desktop/KataGo/My games/yehu-games")
os.makedirs(output_dir, exist_ok=True)

success, skipped, failed = 0, 0, []

for i, game in enumerate(games, 1):
    fn = make_filename(game)
    path = os.path.join(output_dir, fn)
    
    if os.path.exists(path) and os.path.getsize(path) > 0:
        skipped += 1
        continue
    
    try:
        sgf = fetch_sgf(game['chessid'])
        with open(path, "w", encoding="utf-8") as f:
            f.write(sgf)
        print(f"[{i}/{len(games)}] ✓ {fn}")
        success += 1
    except Exception as e:
        print(f"[{i}/{len(games)}] ✗ {game['chessid']}: {e}")
        failed.append(game['chessid'])
    
    time.sleep(0.3)

print(f"\n{'='*50}")
print(f"完成: 成功 {success}, 跳过 {skipped}, 失败 {len(failed)}")
print(f"输出目录: {output_dir}")
```
